```python
import pathlib
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn
from matplotlib.patches import Patch

pathlib.Path("out").mkdir(exist_ok=True)
```

```python
plt.rcParams['font.family'] = 'Arial'
```

```python
def plot_bargraph(diff_value, name, l_filter=None):
    data = {
        'Model': ['ANI1x', 'Classical transfer', 'Classical transfer', 'Classical transfer',
                'Quantum transfer', 'Quantum transfer', 'Quantum transfer'],
        'l': [None, 1, 2, 3, 1, 2, 3],
    }

    # Create a DataFrame
    df = pd.DataFrame(data)
    df[name] = [d * 627.5095 for d in diff_value]

    # Optionally restrict to a subset of l values (ANI1x, which has l=None, is always kept)
    if l_filter is not None:
        df = df[df['l'].isna() | df['l'].isin(l_filter)].reset_index(drop=True)

    # Re-plot with adjusted positions
    fig, ax = plt.subplots(figsize=(8, 6))
    
    colors = {1: 'blue', 2: 'orange', 3: 'green'}
    bar_colors = [colors.get(row['l'], 'gray') for _, row in df.iterrows()]

    # Assign x positions sequentially, leaving a gap whenever the Model changes
    x_positions = []
    pos = 0
    prev_model = None
    for _, row in df.iterrows():
        if prev_model is not None and row['Model'] != prev_model:
            pos += 1
        x_positions.append(pos)
        prev_model = row['Model']
        pos += 1

    bars = ax.bar(x_positions, df[name], color=bar_colors)


    # Set x-axis ticks and labels
    ax.set_xticks(x_positions)
    ax.set_xticklabels([""] * len(x_positions))

    # Add group labels again, computed from the (possibly filtered) bar positions
    group_positions = {}
    for model, xpos in zip(df['Model'], x_positions):
        start, end = group_positions.get(model, (xpos, xpos))
        group_positions[model] = (min(start, xpos), max(end, xpos))

    for label, (start, end) in group_positions.items():
        xpos = (start + end) / 2
        ax.text(xpos, -0.01, label, ha='center', va='top', fontsize=24, transform=ax.get_xaxis_transform())
        #ax.plot([start - 0.4, end + 0.4], [-0.05, -0.05], color='black', transform=ax.get_xaxis_transform(), clip_on=False)

    ax.yaxis.grid(True, linestyle="--", linewidth=0.5)
    # Add legend for the l values actually present in the (possibly filtered) data
    legend_labels = {1: 'tab:blue', 2: 'tab:orange', 3: 'tab:green'}
    present_l = sorted(l for l in df['l'].unique() if pd.notna(l))
    legend_elements = [Patch(facecolor=legend_labels[l], label=f"$l={l}$") for l in present_l]
    ax.legend(handles=legend_elements, fontsize=24)

    # Set axis labels and title
    ax.set_ylabel("Prediction Error (kcal/mol)", fontsize=24)
    ax.set_yticks(np.arange(0, 51, 10))
    ax.tick_params(axis="y", labelsize=24)
    #ax.set_title("Cholesterol Prediction Error")

    # Save the figure
    plt.tight_layout()
    plt.savefig(f"out/{name}_barplot.png")
```

```python
plot_bargraph([0.01274, 0.03926, 0.00474, 0.00326, 0.00826, 0.00274, 0.00726], "cholesterol_4qu", l_filter=[1])
```
